```python
from sklearn.preprocessing import OrdinalEncoder
o=OrdinalEncoder()
df[['tile_id']]=o.fit_transform(df[['tile_id']])
df[['density']]=o.fit_transform(df[['density']])
df[['height']]=o.fit_transform(df[['height']])
df[['settlement_type']]=o.fit_transform(df[['settlement_type']])
df[['roof_material']]=o.fit_transform(df[['roof_material']])
df[['roof_layout']]=o.fit_transform(df[['roof_layout']])
df[['is_gated']]=o.fit_transform(df[['is_gated']])
df[['geometry']]=o.fit_transform(df[['geometry']])
```

```python
df['height']=df['height'].fillna(df['height'].mode()[0])
df['roof_material']=df['roof_material'].fillna(df['roof_material'].median)
df['roof_layout']=df['roof_layout'].fillna(df['roof_layout'].mean())
df['roof_area']=df['roof_area'].fillna(df['roof_area'].median())
```
