```python
from sklearn.preprocessing import OrdinalEncoder
o=OrdinalEncoder()
df[['Employee_Name']]=o.fit_transform(df[['Employee_Name']])
df[['Position']]=o.fit_transform(df[['Position']])
df[['State']]=o.fit_transform(df[['State']])
df[['DOB']]=o.fit_transform(df[['DOB']])
df[['Sex']]=o.fit_transform(df[['Sex']])
df[['MaritalDesc']]=o.fit_transform(df[['MaritalDesc']])
df[['CitizenDesc']]=o.fit_transform(df[['CitizenDesc']])
df[['HispanicLatino']]=o.fit_transform(df[['HispanicLatino']])

df[['RaceDesc']]=o.fit_transform(df[['RaceDesc']])
df[['DateofHire']]=o.fit_transform(df[['DateofHire']])
df[['DateofTermination']]=o.fit_transform(df[['DateofTermination']])
df[['TermReason']]=o.fit_transform(df[['TermReason']])
df[['EmploymentStatus']]=o.fit_transform(df[['EmploymentStatus']])
df[['Department']]=o.fit_transform(df[['Department']])
df[['ManagerName']]=o.fit_transform(df[['ManagerName']])
df[['RecruitmentSource']]=o.fit_transform(df[['RecruitmentSource']])
df[['PerformanceScore']]=o.fit_transform(df[['PerformanceScore']])
df[['LastPerformanceReview_Date']]=o.fit_transform(df[['LastPerformanceReview_Date']])
```

```python
x=df.drop(columns="Absences")
y=df["Absences"]
x
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x=scaler.fit_transform(x)
x
```
