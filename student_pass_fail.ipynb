```python
from sklearn.preprocessing import OrdinalEncoder
o=OrdinalEncoder()
```

```python
df['Study Hours per Week']=df['Study Hours per Week'].fillna(df['Study Hours per Week'].mean())
df['Participation in Extracurricular Activities']=df['Participation in Extracurricular Activities'].fillna(df['Participation in Extracurricular Activities'].mode()[0])
df['Parent Education Level']=df['Parent Education Level'].fillna(df['Parent Education Level'].mode()[0])
df['Passed']=df['Passed'].fillna(df['Passed'].mode()[0])
```

```python
from scipy.stats import chi2_contingency
```

```python
df['Parent Education Level'].value_counts()
```

```python
table = pd.crosstab(df['Parent Education Level'], df['Passed'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
from statsmodels.formula.api import ols
df = pd.read_csv('student_performance_prediction.csv')
df.columns
```

```python
from statsmodels.stats.anova import _get_covariance, anova_lm
```

```python
formula = 'Attendance Rate ~ C(Passed)'
model = ols(formula, df).fit()
aov_table = anova_lm(model)
print(aov_table)
```
