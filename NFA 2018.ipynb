```python
df['forest_land']=df['forest_land'].fillna(df['forest_land'].mean())
df['crop_land']=df['crop_land'].fillna(df['crop_land'].mean())
df['ISO alpha-3 code']=df['ISO alpha-3 code'].fillna(df['ISO alpha-3 code'].mode()[0])
df['grazing_land']=df['grazing_land'].fillna(df['grazing_land'].median())
df['fishing_ground']=df['fishing_ground'].fillna(df['fishing_ground'].median())
df['built_up_land']=df['built_up_land'].fillna(df['built_up_land'].median())
df['carbon']=df['carbon'].fillna(df['carbon'].median())
df['Percapita GDP (2010 USD)']=df['Percapita GDP (2010 USD)'].fillna(df['Percapita GDP (2010 USD)'].median())
```
