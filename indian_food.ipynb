```python
table = pd.crosstab(df['diet'], df['region'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
table = pd.crosstab(df['state'], df['region'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
table = pd.crosstab(df['name'], df['region'])
print(table)

chi2, p, dof, expected = chi2_contingency(table)
alpha = 0.05
print(f'Chi2: {chi2:.2f}, p-value: {p:.4f}, Degrees of freedom: {dof}')
```

```python
from statsmodels.formula.api import ols
df = pd.read_csv('indian_food.csv')
df.columns
```

```python
from statsmodels.stats.anova import _get_covariance, anova_lm
formula = 'prep_time ~ C(region)'
model = ols(formula, df).fit()
aov_table = anova_lm(model)
print(aov_table)
```

```python
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

formula = 'cook_time ~ C(region)'
model = ols(formula, df).fit()
aov_table = anova_lm(model)
print(aov_table)
```
